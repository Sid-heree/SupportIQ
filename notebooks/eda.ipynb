```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Set visualization style
sns.set_theme(style="whitegrid")
plt.rcParams["figure.figsize"] = (10, 6)

# Load the cleaned data
df = pd.read_csv("../data/processed/cleaned_tickets.csv")
print(f"Dataset ready for EDA! Shape: {df.shape}")
```

```python
fig, axes = plt.subplots(1, 3, figsize=(20, 6))

# Priority Distribution
sns.countplot(data=df, x='priority', order=df['priority'].value_counts().index, ax=axes[0], palette="Reds_r")
axes[0].set_title("Ticket Distribution by Priority")
axes[0].tick_params(axis='x', rotation=45)

# Queue Distribution
sns.countplot(data=df, x='queue', order=df['queue'].value_counts().index, ax=axes[1], palette="Blues_r")
axes[1].set_title("Ticket Distribution by Queue")
axes[1].tick_params(axis='x', rotation=45)

# Type Distribution
sns.countplot(data=df, x='type', order=df['type'].value_counts().index, ax=axes[2], palette="Greens_r")
axes[2].set_title("Ticket Distribution by Type")
axes[2].tick_params(axis='x', rotation=45)

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 5))
sns.countplot(data=df, y='language', order=df['language'].value_counts().index, palette="viridis")
plt.title("Ticket Distribution by Language")
plt.xlabel("Number of Tickets")
plt.ylabel("Language Code")
plt.show()
```

```python
# Create a crosstab of Queue and Priority
queue_priority_crosstab = pd.crosstab(df['queue'], df['priority'])

plt.figure(figsize=(12, 8))
sns.heatmap(queue_priority_crosstab, annot=True, fmt="d", cmap="YlGnBu", linewidths=.5)
plt.title("Heatmap: Queue vs. Priority")
plt.ylabel("Queue")
plt.xlabel("Priority")
plt.show()
```
